```python
import pdfplumber

def get_pdf_page(pdf_path, page_number):
    with pdfplumber.open(pdf_path) as pdf:

        if not 1 <= page_number <= len(pdf.pages):
            return "Invalid page number"

        page = pdf.pages[page_number - 1]

        # Extract text preserving layout
        text = page.extract_text(layout=False)

        # Extract tables
        tables = page.extract_tables()

        print("PAGE CONTENT:\n")
        print(text)

        for i, table in enumerate(tables, 1):
            print(f"\nTABLE {i}:")
            for row in table:
                print(row)

        return text
```

```python
import pdfplumber
import html

def get_pdf_page(pdf_path, page_number):

    with pdfplumber.open(pdf_path) as pdf:
        page = pdf.pages[page_number - 1]

        html_content = []

        # Extract text with font information
        lines = page.extract_text_lines()

        for line in lines:
            text = line["text"].strip()

            if not text:
                continue

            # Find maximum font size in the line
            font_size = max(
                char["size"] for char in line["chars"]
            )

            text = html.escape(text)

            if font_size >= 18:
                html_content.append(f"<h1>{text}</h1>")

            elif font_size >= 14:
                html_content.append(f"<h2>{text}</h2>")

            else:
                html_content.append(f"<p>{text}</p>")

        # Extract tables
        for table in page.extract_tables():
            html_content.append("<table border='0'>")

            for row in table:
                html_content.append("<tr>")

                for cell in row:
                    html_content.append(
                        f"<td>{html.escape(cell or '')}</td>"
                    )

                html_content.append("</tr>")

            html_content.append("</table>")

        result = "\n".join(html_content)

        print(result)
        return result
```

```python
import os
import io
import json
import base64
import pdfplumber
import pymupdf

from groq import Groq
from dotenv import load_dotenv
from openpyxl import Workbook
from openpyxl.styles import Alignment

load_dotenv()

client = Groq(api_key=os.environ["GROQ_API_KEY"])

MODEL = "qwen/qwen3.8-27b"


# 1. Find all pages containing tables
def get_pages_with_tables(pdf_path):
    table_pages = []

    with pdfplumber.open(pdf_path) as pdf:
        for page_num, page in enumerate(pdf.pages, start=1):
            if page.find_tables():
                table_pages.append(page_num)

    return table_pages


# 2. Convert a table into a cropped screenshot
def get_table_image(pdf_page, bbox):
    # Add a little padding around the table
    x0, y0, x1, y1 = bbox

    clip = pymupdf.Rect(
        max(0, x0 - 5),
        max(0, y0 - 5),
        min(pdf_page.rect.width, x1 + 5),
        min(pdf_page.rect.height, y1 + 5)
    )

    pix = pdf_page.get_pixmap(
        matrix=pymupdf.Matrix(2, 2),
        clip=clip,
        alpha=False
    )

    image_bytes = pix.tobytes("png")

    return base64.b64encode(image_bytes).decode("utf-8")


# 3. Extract the table using Groq Vision
def extract_table_with_groq(image_base64):

    prompt = """
    You are an expert at extracting tables from PDF images.

    Extract the complete table shown in this image.

    Rules:
    1. Preserve all rows and columns in the correct order.
    2. Preserve the exact content of every cell.
    3. Do not summarize, calculate, or invent values.
    4. Use an empty string for blank cells.
    5. Preserve multi-line text within its original cell.
    6. Do not split one cell into multiple columns.
    7. Return ONLY a JSON object in this format:

    {
        "rows": [
            ["Header 1", "Header 2"],
            ["Value 1", "Value 2"]
        ]
    }
    """

    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {
                "role": "user",
                "content": [
                    {"type": "text", "text": prompt},
                    {
                        "type": "image_url",
                        "image_url": {
                            "url": (
                                "data:image/png;base64,"
                                + image_base64
                            )
                        }
                    }
                ]
            }
        ],
        temperature=0,
        max_completion_tokens=8192,
        response_format={"type": "json_object"}
    )

    result = json.loads(response.choices[0].message.content)

    rows = result.get("rows", [])

    if not isinstance(rows, list):
        raise ValueError("Invalid table JSON returned by Groq")

    return rows


def pdf_tables_to_excel(pdf_path, output_path="HDFC_Tables.xlsx"):

    workbook = Workbook()
    workbook.remove(workbook.active)

    sheet_count = 0

    table_pages = get_pages_with_tables(pdf_path)
    print("Pages with tables:", table_pages)

    with pdfplumber.open(pdf_path) as pdf, pymupdf.open(pdf_path) as doc:

        for page_num in table_pages:
            print(f"\nProcessing page {page_num}")

            plumber_page = pdf.pages[page_num - 1]
            image_page = doc[page_num - 1]

            tables = plumber_page.find_tables()

            tables = sorted(
                tables,
                key=lambda t: (round(t.bbox[1] / 10), t.bbox[0])
            )

            for table_num, table in enumerate(tables, start=1):

                sheet_name = f"Page_{page_num}_Table_{table_num}"

                try:
                    image = get_table_image(image_page, table.bbox)

                    rows = extract_table_with_groq(image)

                    if not rows:
                        print(f"Empty result: {sheet_name}")
                        continue

                    if not all(isinstance(row, list) for row in rows):
                        raise ValueError("Invalid table rows returned by Groq")

                    sheet = workbook.create_sheet(title=sheet_name)

                    for row in rows:
                        # Protect against interpreting PDF text as formulas
                        safe_row = [
                            str(value) if value is not None else ""
                            for value in row
                        ]

                        safe_row = [
                            "'" + value if value.startswith(
                                ("=", "+", "-", "@")
                            ) else value
                            for value in safe_row
                        ]

                        sheet.append(safe_row)

                    # Formatting
                    for row in sheet.iter_rows():
                        for cell in row:
                            cell.alignment = Alignment(
                                wrap_text=True,
                                vertical="top"
                            )

                    # Adjust column widths
                    for column in sheet.columns:
                        letter = column[0].column_letter
                        sheet.column_dimensions[letter].width = 25

                    sheet_count += 1
                    print(f"Created: {sheet_name}")

                except Exception as e:
                    print(f"Failed {sheet_name}: {e}")

    if sheet_count == 0:
        print("No tables were successfully extracted.")
        return None

    workbook.save(output_path)

    print(f"\nSaved {sheet_count} tables to {output_path}")

    return output_path
```

```python
pdf_path = r"D:\Mission_Jump\Prep\gen_ai_learned\HDFC_OptimaSecure_Policy_Doc.pdf"

output = pdf_tables_to_excel(
    pdf_path,
    output_path=r"D:\Mission_Jump\Prep\gen_ai_learned\HDFC_Tables.xlsx"
)
```
